```python
import os
import shutil
from PyPDF2 import PdfReader
```

```python
subjects = {
    "Python": [
        "python", "numpy", "pandas", "list", "tuple", "dictionary", "function"
    ],

    "DBMS": [
        "sql", "database", "query", "normalization", "primary key", "foreign key", "transaction"
    ],

    "Cloud": [
        "cloud", "aws", "azure", "virtual machine", "storage", "iaas", "paas", "saas"
    ],

    "Machine_Learning": [
        "machine learning", "regression", "classification", "dataset", "model", "training", "prediction"
    ],

    "NLP": [
        "natural language", "nlp", "tokenization", "stemming", "sentiment", "text", "stopwords"
    ],

    "Computer_Networks": [
        "tcp", "udp", "ip address", "router", "network", "osi", "protocol"
    ]
}
print("Subjects configured successfully!")
```

```python
# Folder that contains the loose PDF files to be organized
source_folder = "."

# Folder where the subject-wise sub-folders will be created
target_folder = "Study_Materials"

os.makedirs(target_folder, exist_ok=True)

print(f"Reading PDFs from: {os.path.abspath(source_folder)}")
print(f"Organizing them into: {os.path.abspath(target_folder)}")
```

```python
def extract_text_from_pdf(file_path):
    
    text = ""

    try:
        reader = PdfReader(file_path)

        for page in reader.pages:
            page_text = page.extract_text()

            if page_text:
                text += page_text.lower() + " "

    except Exception as e:
        print("Error reading:", file_path)
        print(e)

    return text
```

```python
def detect_subject(text):

    scores = {}

    for subject, keywords in subjects.items():

        score = 0

        for keyword in keywords:

            if keyword.lower() in text:
                score += 1

        scores[subject] = score

    best_subject = max(scores, key=scores.get)

    if scores[best_subject] == 0:
        return "Others"

    return best_subject
```

```python
files = os.listdir(source_folder)

moved = 0

for file in files:

    if not file.lower().endswith(".pdf"):
        continue

    file_path = os.path.join(source_folder, file)

    # Skip anything that is not an actual file (e.g. folders)
    if not os.path.isfile(file_path):
        continue

    # Extract PDF content
    text = extract_text_from_pdf(file_path)

    if not text.strip():
        print(f"{file}  ->  (skipped: no readable text, maybe a scanned PDF)")
        continue

    # Detect subject
    subject = detect_subject(text)

    # Create destination folder inside Study_Materials
    destination_folder = os.path.join(target_folder, subject)
    os.makedirs(destination_folder, exist_ok=True)

    # Move file
    destination = os.path.join(destination_folder, file)
    shutil.move(file_path, destination)
    moved += 1

    print(f"{file}  ->  {subject}")

print(f"\n{moved} file(s) organized successfully!")
```

```python
print("Final folder structure:\n")

for root, folders, files in os.walk(target_folder):

    level = root.replace(target_folder, "").count(os.sep)

    indent = "    " * level

    print(indent + os.path.basename(root) + "/")

    for file in files:
        print(indent + "    " + file)
```
