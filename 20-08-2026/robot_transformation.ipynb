# Experiment 2: 2D Rigid Body Frame Transformations & Spatial Visualization

This notebook implements the 2D homogeneous transformation from the provided lab exercise.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 1. Define original point in 2D homogeneous coordinates [x, y, 1.0]
point_global = np.array([2.0, -1.0, 1.0])

# 2. Transformation parameters
theta = np.radians(45)  # Rotate 45 degrees
tx, ty = 3.0, 2.0       # Translate +3 in X, +2 in Y

# 3. Construct 3x3 Homogeneous Transformation Matrix T
T = np.array([
    [np.cos(theta), -np.sin(theta), tx],
    [np.sin(theta),  np.cos(theta), ty],
    [0,              0,              1]
])

# 4. Compute transformed point using @ matrix multiplication
transformed_point = T @ point_global

print('Transformation Matrix T:')
print(T)
print('\nOriginal point:', point_global)
print('Transformed point:', transformed_point)
```

```python
# 5. Visualization
plt.figure(figsize=(6, 6))

# World Origin
plt.scatter(0, 0, color='black', s=100, label='World Origin')

# Original Position
plt.scatter(point_global[0], point_global[1], color='blue', s=100,
            label='Original Position')
plt.plot([0, point_global[0]], [0, point_global[1]], 'b--')

# Transformed Position
plt.scatter(transformed_point[0], transformed_point[1], color='red', s=100,
            label='Transformed Position')
plt.plot([0, transformed_point[0]], [0, transformed_point[1]], 'r--')

# Plot formatting
plt.xlim(-1, 7)
plt.ylim(-1, 7)
plt.grid(True)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('2D Rigid Body Frame Transformation')
plt.legend()
plt.gca().set_aspect('equal', adjustable='box')
plt.show()
```
